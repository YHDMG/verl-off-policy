# Experiment: Math Entropy Study

This notebook is the interactive entry point for the `math_entropy_analysis` pipeline.

Workflow:
- load `artifacts/<run_name>/generations.parquet`, `token_metrics.parquet`, and `sequence_features.parquet`
- compare correct vs incorrect samples
- inspect entropy trajectories and answer-span uncertainty
- render a single-sample entropy heatmap
- regenerate summary plots for the current run


```python
from __future__ import annotations

from pathlib import Path
from statistics import fmean

from math_entropy_analysis.dataset import load_records
from math_entropy_analysis.plots import run_plotting
from math_entropy_analysis.visualize import write_entropy_heatmap

RUN_DIR = Path('artifacts') / 'example_run'
RUN_DIR
```

## Load artifacts

Point `RUN_DIR` at a completed run before executing the cells below.


```python
generations = load_records(RUN_DIR / 'generations.parquet')
token_metrics = load_records(RUN_DIR / 'token_metrics.parquet')
sequence_features = load_records(RUN_DIR / 'sequence_features.parquet')

summary = {
    'num_generations': len(generations),
    'num_token_rows': len(token_metrics),
    'num_feature_rows': len(sequence_features),
    'accuracy': fmean([1.0 if row.get('is_correct') else 0.0 for row in sequence_features]) if sequence_features else None,
}
summary
```

```python
correct = [row for row in sequence_features if row.get('is_correct')]
incorrect = [row for row in sequence_features if not row.get('is_correct')]

comparison = {
    'correct_mean_entropy': fmean([row['mean_entropy'] for row in correct if row.get('mean_entropy') is not None]) if correct else None,
    'incorrect_mean_entropy': fmean([row['mean_entropy'] for row in incorrect if row.get('mean_entropy') is not None]) if incorrect else None,
    'correct_answer_span_entropy': fmean([row['answer_span_entropy_mean'] for row in correct if row.get('answer_span_entropy_mean') is not None]) if correct else None,
    'incorrect_answer_span_entropy': fmean([row['answer_span_entropy_mean'] for row in incorrect if row.get('answer_span_entropy_mean') is not None]) if incorrect else None,
}
comparison
```

```python
sample_id = str(generations[0]['sample_id'])
decode_id = int(generations[0]['decode_id'])
heatmap_path = write_entropy_heatmap(
    token_metrics_path=RUN_DIR / 'token_metrics.parquet',
    generations_path=RUN_DIR / 'generations.parquet',
    sample_id=sample_id,
    decode_id=decode_id,
    output_path=RUN_DIR / 'plots' / f'{sample_id}_{decode_id}_entropy_heatmap.html',
)
heatmap_path
```

```python
plots_dir = run_plotting(
    sequence_features_path=RUN_DIR / 'sequence_features.parquet',
    token_metrics_path=RUN_DIR / 'token_metrics.parquet',
    output_dir=RUN_DIR / 'plots',
)
sorted(path.name for path in Path(plots_dir).glob('*.png'))
```
